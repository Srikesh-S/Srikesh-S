# Employee Data Analysis
## Week 1 – Day 5 | Task 1
Analyze a sample of 10 employee records using Pandas, Matplotlib, and Seaborn. Charts are saved to `figures/`.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('employee_data.csv')
os.makedirs('figures', exist_ok=True)
sns.set_theme(style='whitegrid')
colors = ['#4472C4', '#ED7D31', '#70AD47', '#A5A5A5']
print(df.head(10))
print('Rows:', len(df))
```

## Chart 1 — Bar chart: Employee count by department
**Insight:** IT has the largest headcount in this sample (4 employees); other departments have 2 each.

```python
department_count = df['Department'].value_counts()
plt.figure(figsize=(8,5))
department_count.plot(kind='bar', color=colors[0])
plt.title('Employee Count by Department')
plt.xlabel('Department')
plt.ylabel('Employees (count)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('figures/bar_chart.png', dpi=300)
plt.show()
```

## Chart 2 — Histogram: Salary distribution
**Insight:** Monthly salaries range from ₹25,000 to ₹45,000 in this sample.

```python
plt.figure(figsize=(8,5))
plt.hist(df['Salary'], bins=5, color=colors[1], edgecolor='black')
plt.title('Employee Salary Distribution')
plt.xlabel('Monthly Salary (INR)')
plt.ylabel('Employees (count)')
plt.tight_layout()
plt.savefig('figures/histogram.png', dpi=300)
plt.show()
```

## Chart 3 — Scatter plot: Age vs salary
**Insight:** This sample suggests a general upward pattern, but age alone does not determine salary.

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Age'], df['Salary'], color=colors[0], alpha=0.8, label='Employees')
plt.title('Age vs Salary Analysis')
plt.xlabel('Age (years)')
plt.ylabel('Monthly Salary (INR)')
plt.legend()
plt.tight_layout()
plt.savefig('figures/scatter_plot.png', dpi=300)
plt.show()
```

## Chart 4 — Line chart: Monthly joining trend
**Insight:** One employee joined in each month from January to September 2025 in this sample.

```python
df['JoiningDate'] = pd.to_datetime(df['JoiningDate'])
monthly_joining = df.groupby(df['JoiningDate'].dt.to_period('M')).size()
monthly_joining.index = monthly_joining.index.astype(str)
plt.figure(figsize=(9,5))
plt.plot(monthly_joining.index, monthly_joining.values, marker='o', color=colors[0], label='Employees joined')
plt.title('Monthly Employee Joining Trend')
plt.xlabel('Month')
plt.ylabel('New joiners (count)')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.savefig('figures/line_chart.png', dpi=300)
plt.show()
```

## Chart 5 — Box plot: Salary by department
**Insight:** IT salaries are higher in this sample; department groups are small, so this is illustrative only.

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Department', y='Salary', color=colors[2])
plt.title('Salary Comparison Across Departments')
plt.xlabel('Department')
plt.ylabel('Monthly Salary (INR)')
plt.tight_layout()
plt.savefig('figures/box_plot.png', dpi=300)
plt.show()
```

## Summary
- **Dataset:** 10 synthetic employee records.
- **Findings:** IT has 4 employees; salaries range ₹25,000–₹45,000 per month; one joiner per month from Jan–Sep 2025.
- **Recommendations:** Review staffing needs by department; assess compensation using role and experience as well as department.
- **Limitations:** Small synthetic sample; no role, experience, performance, or attrition data.
