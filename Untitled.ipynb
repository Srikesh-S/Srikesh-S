```python
import time

elements = list(range(1000000))

def test_loop():
    results = []
    for x in elements:
        results.append(x * 2)

print("Testing For Loop:")
%timeit test_loop()

print("\nTesting List Comprehension:")
%timeit [x * 2 for x in elements]
```
